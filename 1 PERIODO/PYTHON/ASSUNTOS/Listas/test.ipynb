Exercicio de Slides

Questão 1

```python
L = [5, 6, 4, 3, 10, 16, 2, 1]
maior = L[0]
menor = L[0]
for i in L:
    if i > maior:
        maior = i
    if i < menor:
        menor = i

print(maior)
print(menor)
```

Questão 2

```python
T = [ -10, -8, 0, 1, 2, 5, -2, -4]
maior = T[0]
menor = T[0]
soma = 0
for i in T:
    if i > maior:
        maior = i
    if i < menor:
        menor = i

for j in T:
    soma += j

print("| Maior:", maior, "| Menor: ", menor, "| Média:", soma/len(T))
```

Questão 3

```python
Lista = []

while True:
    lisr = int(input("Digite um número para incrementar(0 sai): "))
    if lisr == 0:
        break
    Lista.append(lisr)

listaimpar = []
listapar = []

for i in Lista:
    if i % 2 == 0:
        listapar.append(i)
    else:
        listaimpar.append(i)

print(listaimpar)
print(listapar)
```
