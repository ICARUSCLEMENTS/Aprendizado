```python
tabela = {
   "Alface": 0.45,
   "Batata": 1.20,
   "Tomate": 2.30,
   "Feijão": 1.50
}

while True:
    produto = input("Digite o nome do produto, fim para terminar: ").capitalize()
    if produto == "Fim":
        break
    if produto in tabela:
        print(f"Preço {tabela[produto]:.2f}")
    else:
        print("Produto não encontrado")
print("Até mais")
```

```python


mercado = {
    "maça" : 1.50,
    "banana" : 0.80,
    "pão" : 3.00,
    "leite" : 2.50,
    "arroz" : 5.00
}

print("Bem-vindo ao mercado")

somatorio = 0
while True:
    print  ("-" * 20)
    print("Produtos disponíveis: ")

    for i in mercado:
        print(f"{i} | {mercado[i]:.2f}")
    produto = input("Digite o nome do produto que deseja, fim para terminar: ").lower()

    if produto == "fim":
        break
    
    elif produto in mercado:
        print(f"Preço: {mercado[produto]:.2f}")
        quantidade = int(input("Digite a quantidade que deseja: "))
    
        if quantidade < 0:
            print("Quantidade inválida")
            continue
    
        elif quantidade == 0:
            continue
    
        else:
            print(f"Total: {mercado[produto] * quantidade:.2f}")
        
        somatorio += mercado[produto] * quantidade
    
    
    else:
        print("Produto não encontrado")

print("-" * 20)
print(f"Total da compra: {somatorio:.2f}")
print("-" * 20)
```
